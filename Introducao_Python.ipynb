## Laços de Repetição - While e For

```python
i = 0

while Aprovado:
    Nota_SABER = Notas_SABER[i]

    if Nota_SABER >= 10:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Excepcional \n")
    elif Nota_SABER >= 9.5:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Ideal \n")
    elif Nota_SABER >= 9.0:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Meta \n")
    else:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Reprovado \n")
        # Aprovado = False
        # break

    i += 1
    if i >= len(Notas_SABER):
        break

```
