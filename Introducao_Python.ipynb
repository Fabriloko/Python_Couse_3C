```python
import math
import pandas as pd
from io import StringIO

# 1. Exibir informações na tela
print("Olá! Este é um exemplo de introdução ao Python.")
print("A soma de 2 + 3 é:", 2 + 3)

# 2. Criar e ler um arquivo de texto
with open("exemplo.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("Primeira linha do arquivo.\nSegunda linha do arquivo.")

with open("exemplo.txt", "r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()

print("\nConteúdo do arquivo:")
print(conteudo)

# 3. Usar funções matemáticas

print("\nFunções matemáticas:")
print("Raiz quadrada de 25:", math.sqrt(25))
print("Potência de 2 elevado a 3:", math.pow(2, 3))
print("Valor de pi:", math.pi)
print("Arredondamento para cima de 4.2:", math.ceil(4.2))

# 4. Criar e explorar um DataFrame do Pandas

dados = {
    "Nome": ["Ana", "Bruno", "Carla"],
    "Idade": [23, 31, 27],
    "Nota": [8.5, 7.0, 9.2],
}

df = pd.DataFrame(dados)

print("\nDataFrame:")
print(df)

print("\nMédia das notas:", df["Nota"].mean())
print("\nPessoas com nota maior que 8:")
print(df[df["Nota"] > 8])

# 5. Ler dados de um arquivo CSV

conteudo_csv = """Produto,Preço
Caderno,12.50
Caneta,3.00
Livro,45.90
"""

df_csv = pd.read_csv(StringIO(conteudo_csv))
print("\nDataFrame lido do CSV:")
print(df_csv)
```
