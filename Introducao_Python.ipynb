# Introdução a Linguagem Python

## Variáveis

```python
# Texto (String)
Nome_Completo = "Fabricio Araujo Barbosa"
# Inteiro (Int)
Idade = 26
# Flutuante/Decimal (Float)
Nota_SABER = 9.5
Altura = 1.63
Peso = 64.55
# Verdadeiro/Falso (Boolean)
Aprovado = True
```

## Funções

```python
#Printando os tipos de dados
print(type(Nome_Completo))
print(type(Idade))
print(type(Nota_SABER))
print(type(Aprovado))
```

```python
Ano_Nascimento = 2026 - Idade
print(f'Ano de Nascimento: {Ano_Nascimento}')

IMC = Peso / (Altura ** 2)
print(f'IMC: {IMC:.2f}')
```

## Condicionais - If, Elif, Else

```python
if Nota_SABER >= 10:
    print("Excepcional")
elif Nota_SABER >= 9.5:
    print("Ideal")
elif Nota_SABER >= 9.0:
    print("Meta")
else:
    print("Reprovado")
```

## Tipo de Variável: Listas

```python
Notas_SABER = [10, 9.5, 9.0, 8.5, 8.0, 10, 8.0, 9.5, 9.0, 8.5, 8.0, 10]

print('Tamanho da Lista:', len(Notas_SABER))
print(f'Tipo da Lista: {type(Notas_SABER)}')
```

## Laços de Repetição - While e For

```python

while Aprovado:
    print("Parabéns, você atingiu a meta!")

    if Nota_SABER >= 10:
        print("Excepcional")
    elif Nota_SABER >= 9.5:
        print("Ideal")
    elif Nota_SABER >= 9.0:
        print("Meta")
    else:
        print("Reprovado")
    break

```
