# Introdução a Linguagem Python

## Variáveis

```python
# Texto (String)
Nome_Completo = "Fabricio Araujo Barbosa"
# Inteiro (Int)
Idade = 26
# Flutuante/Decimal (Float)
Altura = 1.63
Peso = 64.55
# Verdadeiro/Falso (Boolean)
Aprovado = True
```

## Operadores  

### Aritméticos

```python
# Operadores aritméticos
x = 10
y = 6

print("Soma:", x + y)
print("Subtração:", x - y)
print("Multiplicação:", x * y)
print("Divisão:", x / y)
print("Resto:", x % y)
print("Potência:", x ** y)
print("Raiz quadrada:", x ** ( 1/2 ))
```

### Lógicos

```python
# Operadores logicos
x = 10
y = 6

print("Maior que:", x > y)
print("Menor que:", x < y)
print("Maior ou igual que:", x >= y)
print("Menor ou igual que:", x <= y)
print("Igual a:", x == y)
print("Diferente de:", x != y)
```

### Relacionais

```python
# Operadores relacionais
x = 10
y = 6

print("E:", x > 5 and y < 10) # &&
print("Ou:", x > 5 or y < 10) # !!
print("Negação:", not (x > 5)) # !
```

## Funções

```python
#Printando os tipos de dados
print(type(Nome_Completo))
print(type(Idade))
print(type(Altura))
print(type(Peso))
print(type(Aprovado))
```

```python
# Funções basicas
print(max(Idade, 5)) # Maior valor
print(min(Idade, 5)) # Menor valor
print(f"Variável: {Peso} \nAredondamento: {round(Peso, 1)}") # Arredondamento
print(abs(-10)) # Valor absoluto
print(pow(Altura, 2)) # Potência
print(divmod(10, 3)) # Divisão e resto
```

```python
def Calc_Ano_Nascimento(Idade):
    return 2026 - Idade

def Calc_IMC(Peso, Altura):
    return Peso / (Altura ** 2)
```

## Condicionais - If, Elif, Else

## Tipo de Variável: Listas

```python
Notas_SABER = [10, 9.5, 9.0, 8.5, 8.0, 10, 8.0, 9.5, 9.0, 8.5, 8.0, 10]

print('Tamanho da Lista:', len(Notas_SABER))
print(f'Tipo da Variável: {type(Notas_SABER)}')
```

```python
import numpy as np

Notas_SABER_2 = np.array(Notas_SABER)
print('Tamanho do Array:', Notas_SABER_2.size)
print(f'Tipo da Variável: {type(Notas_SABER_2)}')
print(f'Forma do Array: {Notas_SABER_2.shape}')
```

## Laços de Repetição - While e For

```python
i = 0
Aprovado = True

while Aprovado:
    Nota_SABER = Notas_SABER[i]
    Media_SABER = np.mean(Notas_SABER_2[:i+1])  # Calando a média das notas até o índice atual

    if Nota_SABER >= 10:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print(f'Acumulado nota SABER: {Media_SABER:.2f}')
        print("Excepcional \n")
    elif Nota_SABER >= 9.5:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print(f'Acumulado nota SABER: {Media_SABER:.2f}')
        print("Ideal \n")
    elif Nota_SABER >= 9.0:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print(f'Acumulado nota SABER: {Media_SABER:.2f}')
        print("Meta \n")
    else:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print(f'Acumulado nota SABER: {Media_SABER:.2f}')
        print("Reprovado \n")
        # Aprovado = False
        # break

    i += 1
    if i >= len(Notas_SABER):
        break

```

```python
for i in range(len(Notas_SABER)):
    Nota_SABER = Notas_SABER[i]
    Media_SABER = np.mean(Notas_SABER_2[:i+1])  # Calando a média das notas até o índice atual


    if Nota_SABER >= 10:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print(f'Acumulado nota SABER: {Media_SABER:.2f} \n')
    elif Nota_SABER >= 9.5:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print(f'Acumulado nota SABER: {Media_SABER:.2f} \n')
    elif Nota_SABER >= 9.0:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print(f'Acumulado nota SABER: {Media_SABER:.2f} \n')
    else:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print(f'Acumulado nota SABER: {Media_SABER:.2f} \n')
        # Aprovado = False
        # break

    i += 1
    if i >= len(Notas_SABER):
        break

```

```python
for i, Nota_SABER in enumerate(Notas_SABER):
    Media_SABER = np.mean(Notas_SABER_2[:i+1])  # Calando a média das notas até o índice atual

    if Nota_SABER >= 10:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Excepcional")
    elif Nota_SABER >= 9.5:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Ideal")
    elif Nota_SABER >= 9.0:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Meta")
    else:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Reprovado")
        # Aprovado = False
        # break

    print(f'Acumulado nota SABER: {Media_SABER:.2f} \n')
```

## Estruturas de Dados

```python
# Estuturas basicas de dados

# Lista 
print("Lista de Notas:", Notas_SABER)
print(f'Tipo da Variável: {type(Notas_SABER)} \n')

# Numpy Array (Matriz)
Notas_SABER_Array = np.array(Notas_SABER)
print("Array de Notas:", Notas_SABER_Array)
print(f'Tipo da Variável: {type(Notas_SABER_Array)} \n')

# Tupla
Notas_SABER_Tupla = tuple(Notas_SABER)
print("Tupla de Notas:", Notas_SABER_Tupla) 
print(f'Tipo da Variável: {type(Notas_SABER_Tupla)} \n')

# Set (Conjuntos)
Notas_SABER_Set = set(Notas_SABER)
print("Set de Notas:", Notas_SABER_Set)
print(f'Tipo da Variável: {type(Notas_SABER_Set)} \n')

# Dict (Dicionario)
Notas_SABER_Dict = {f'Mes {i}': nota for i, nota in enumerate(Notas_SABER)}
print("Dicionário de Notas:", Notas_SABER_Dict)
print(f'Tipo da Variável: {type(Notas_SABER_Dict)} \n')
```

## Manipulação de Arquivos

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url)
# Para Excel, use: pd.read_excel('seu_arquivo.xlsx')

print("--- PRIMEIRAS LINHAS ---")
display(df.head())

print("\n--- INFORMAÇÕES E TIPOS DE DADOS ---")
print(df.info())

print("\n--- RESUMO ESTATÍSTICO ---")
display(df.describe())
```

```python
# inserir linha no df
df_new = pd.merge(df, pd.DataFrame([{
    'species': 'Teste',
    'island': 'Torgersen',
    'bill_length_mm': 39.1,
    'bill_depth_mm': 18.7,
    'flipper_length_mm': 181,
    'body_mass_g': 3750,
    'sex': 'MALE'
}]))

# salvando o DataFrame atualizado em um novo arquivo CSV
df_new.to_csv('penguins_updated.csv', index=False)

```
