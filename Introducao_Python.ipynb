# Introdução a Linguagem Python

## Variáveis

```python
# Texto (String)
Nome_Completo = "Fabricio Araujo Barbosa"
# Inteiro (Int)
Idade = 26
# Flutuante/Decimal (Float)
Altura = 1.63
Peso = 64.55
# Verdadeiro/Falso (Boolean)
Aprovado = True
```

## Funções

```python
#Printando os tipos de dados
print(type(Nome_Completo))
print(type(Idade))
print(type(Altura))
print(type(Peso))
print(type(Aprovado))
```

```python
Ano_Nascimento = 2026 - Idade
print(f'Ano de Nascimento: {Ano_Nascimento}')

IMC = Peso / (Altura ** 2)
print(f'IMC: {IMC:.2f}')
```

## Recebendo Informacao do Usuario

```python
Mes_Nascimento = input("Digite o mês de nascimento: ")
print(type(Mes_Nascimento))

Mes_Nascimento = int(Mes_Nascimento)
print(type(Mes_Nascimento))
```

## Condicionais - If, Elif, Else

```python
Mes_Atual = 10

if Mes_Nascimento > 12:
    print("Mês inválido")
elif Mes_Nascimento < 1:
    print("Mês inválido")
else:
    print("Mês válido")
    if Mes_Nascimento <= Mes_Atual:
        Ano_Nascimento = 2026 - Idade
        print(f"Ano de nascimento: {Ano_Nascimento}")
    else:
        Ano_Nascimento = 2026 - Idade - 1
        print(f"Ano de nascimento: {Ano_Nascimento}")
```

## Tipo de Variável: Listas

```python
Notas_SABER = [10, 9.5, 9.0, 8.5, 8.0, 10, 8.0, 9.5, 9.0, 8.5, 8.0, 10]

print('Tamanho da Lista:', len(Notas_SABER))
print(f'Tipo da Variável: {type(Notas_SABER)}')
```

## Laços de Repetição - While e For

```python
i = 0

while Aprovado:
    Nota_SABER = Notas_SABER[i]

    if Nota_SABER >= 10:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Excepcional \n")
    elif Nota_SABER >= 9.5:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Ideal \n")
    elif Nota_SABER >= 9.0:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Meta \n")
    else:
        print(f'Mes: {i}')
        print(f'Nota no SABER: {Nota_SABER}')
        print("Reprovado \n")
        # Aprovado = False
        # break

    i += 1
    if i >= len(Notas_SABER):
        break

```
