# Introdução à Programação Orientada a Objeto

```python
class Animal:
    def __init__(self, name, species, years_old):
        self.name = name
        self.species = species
        self.years_old = years_old

    def speak(self):
        if self.species == "dog":
            return "Woof!"
        elif self.species == "cat":
            return "Meow!"
        elif self.species == "cow":
            return "Moo!"   
        elif self.species == "bird":
            return "Chirp!"
        elif self.species == "fish":
            return "Blub!"
        elif self.species == "horse":
            return "Neigh!"
        else:
            raise NotImplementedError("Subclasses must implement this method.")
        
    def age_in_human_years(self):
        if self.species == "dog":
            return self.years_old * 7
        elif self.species == "cat":
            return self.years_old * 6
        elif self.species == "cow":
            return self.years_old * 4
        elif self.species == "bird":
            return self.years_old * 5
        elif self.species == "fish":
            return self.years_old * 3
        elif self.species == "horse":
            return self.years_old * 5
        else:
            raise NotImplementedError("Subclasses must implement this method.")

    def is_old(self):
        if self.species == "dog":
            if self.years_old > 10:
                return "Really Old"
            elif self.years_old > 7:
                return "Old"
            else:
                return "Young"
        elif self.species == "cat":
            if self.years_old > 12:
                return "Really Old"
            elif self.years_old > 9:
                return "Old"
            else:
                return "Young"
        elif self.species == "cow":
            if self.years_old > 15:
                return "Really Old"
            elif self.years_old > 10:
                return "Old"
            else:
                return "Young"
        elif self.species == "bird":
            if self.years_old > 8:
                return "Really Old"
            elif self.years_old > 4:
                return "Old"
            else:
                return "Young"
        elif self.species == "fish":
            if self.years_old > 5:
                return "Really Old"
            elif self.years_old > 2:
                return "Old"
            else:
                return "Young"
        elif self.species == "horse":
            if self.years_old > 20:
                return "Really Old"
            elif self.years_old > 15:
                return "Old"
            else:
                return "Young"
        else:
            raise NotImplementedError("Subclasses must implement this method.")

    def moviment(self):
        if self.species == "dog":
            return "runs"
        elif self.species == "cat":
            return "slinks"
        elif self.species == "cow":
            return "walks"
        elif self.species == "bird":
            return "flies"
        elif self.species == "fish":
            return "swims"
        elif self.species == "horse":
            return "gallops"
        else:
            raise NotImplementedError("Subclasses must implement this method.")
        
    def species_info(self):
        return f"{self.name} is a {self.species}."
```

```python
def main():
    dog = Animal("Buddy", "dog", 5)
    cat = Animal("Whiskers", "cat", 3)
    cow = Animal("Bessie", "cow", 4)
    bird = Animal("Tweety", "bird", 2)
    fish = Animal("Nemo", "fish", 1)
    horse = Animal("Spirit", "horse", 7)

    animals = [dog, cat, cow, bird, fish, horse]

    for animal in animals:
        print(animal.species_info())
        print(f"{animal.name} says: {animal.speak()}")
        print(f"{animal.name} is {animal.age_in_human_years()} years old in human years and is {animal.is_old()} animal.")
        print(f"{animal.name} moves by: {animal.moviment()}")
        print()
        
if __name__ == "__main__":
    main()
```
